# THE PROBLEM

## 13 minutes
**AVERAGE DELAY IN AMBULANCE RESPONSE TIMES**

Legacy infrastructure cannot handle the low staffing and high call volume especially in underserved boroughs.

> "PEOPLE ARE DYING BECAUSE THERE ARE NO STAFF," 
> - *Anthony Almojera, a lieutenant paramedic recognising the ever increasing callout time.*

### MANUAL RESOURCE ALLOCATION
* Human error in high-stress routing decisions.
* Non-optimised resource allocation adds to an ever increasing response time.

# THE TARGET DEMOGRAPHIC

**PRIMARY ADOPTION FOCUSED ON DENSE METROPOLITAN REGIONS. MODEL TRAINED ON NEW YORK CITY AS IT HAS THE LARGEST AND MOST INFORMATIVE DATASET AVAILABLE.**

* **490,000** VEHICLES 
* **8.8M** NYC POPULATION


**Notes:**
Why New York? Key Drivers
* Traffic Gridlock: 490,000 cars & 300,000 pedestrians flood Midtown daily, blocking response corridors.
* Staffing Shortages: Unprecedented drop in active EMTs creating severe delays.
* System Strain: Union leaders warn the EMS infrastructure is at breaking point.


# THE SOLUTION

### DETERMINING PRIORITY LOCATIONS
Analysis of NYC Open Data EMS Incident Dispatch Data 

### PREDICTIVE ROUTING
AI-driven path optimization.

### WEATHER RELATED INCIDENT PREDICTION
Analyse weather reports to determine preventative action.
